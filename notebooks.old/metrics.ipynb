```python
def normalise_value(column, value):
    """
    Normalise a trait value for comparison.
    """

    if pd.isna(value) or str(value).strip() == "":
        return None

    value = str(value).strip()

    # Dimensional/numeric fields, e.g.
    # plant min. height [m], seed max. length [mm]
    if re.search(r"\[[^\]]+\]", column):
        numbers = re.findall(r"-?\d+(?:\.\d+)?", value)

        if not numbers:
            return None

        return tuple(float(x) for x in numbers)

    # Categorical values:
    # treat comma-separated values as an unordered set
    return frozenset(
        x.strip().lower()
        for x in value.split(",")
        if x.strip()
    )


def compare_rows(pred, eval_, key):
    """
    Compare prediction and evaluation rows for a given key.
    Returns only fields whose normalised values differ.
    """

    pred_matches = pred.loc[pred["key"] == key]
    eval_matches = eval_.loc[eval_["key"] == key]

    if len(pred_matches) == 0:
        raise ValueError(f"No prediction found for key: {key}")

    if len(eval_matches) == 0:
        raise ValueError(f"No evaluation row found for key: {key}")

    pred_row = pred_matches.iloc[0]
    eval_row = eval_matches.iloc[0]

    differences = []

    columns = pred.columns.intersection(eval_.columns)

    for column in columns:
        if column == "key" or column.startswith("Unnamed:"):
            continue

        pred_value = pred_row[column]
        eval_value = eval_row[column]

        pred_norm = normalise_value(column, pred_value)
        eval_norm = normalise_value(column, eval_value)

        if pred_norm != eval_norm:
            differences.append({
                "field": column,
                "Pred": pred_value,
                "Eval": eval_value,
            })

    return pd.DataFrame(differences)

key = "agrostis stolonifera|efloras.flora_of_pakistan"

differences = compare_rows(pred_df, eval_df, key)

differences
```

```python
def compare_all_rows(pred, eval_):
    """
    Compare all matching rows in pred and eval.
    Returns one row per differing trait, including taxon and key.
    """

    differences = []

    # Only compare records present in both datasets
    shared_keys = set(pred["key"]) & set(eval_["key"])

    for key in shared_keys:
        pred_row = pred.loc[pred["key"] == key].iloc[0]
        eval_row = eval_.loc[eval_["key"] == key].iloc[0]

        taxon = eval_row["taxon"]

        for column in pred.columns.intersection(eval_.columns):

            if column in {"key", "taxon", "source", "source_id"}:
                continue

            if column.startswith("Unnamed:"):
                continue

            pred_value = pred_row[column]
            eval_value = eval_row[column]

            pred_norm = normalise_value(column, pred_value)
            eval_norm = normalise_value(column, eval_value)

            if pred_norm != eval_norm:
                differences.append({
                    "taxon": taxon,
                    "key": key,
                    "field": column,
                    "Pred": pred_value,
                    "Eval": eval_value,
                })

    return pd.DataFrame(differences)
```
